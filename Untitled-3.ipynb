```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

print("Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv("/Users/ms/Desktop/SAM AI INTERNSHIP/BostonHousing.csv")

print("Dataset loaded successfully!")
```

```python
# Display the first five rows
df.head()
```

```python
# Check number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display information about the dataset
df.info()
```

```python
# Display descriptive statistics
df.describe()
```

```python
# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Fill missing numerical values with the median
for column in df.select_dtypes(include=np.number).columns:
    df[column] = df[column].fillna(df[column].median())

# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("Missing values after preprocessing:")
print(df.isnull().sum())
```

```python
# Select features for clustering
features = [
    "rm",
    "lstat",
    "ptratio",
    "crim"
]

X = df[features]

print("Features selected for clustering:")
print(features)

X.head()
```

```python
# Standardize the features
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Features standardized successfully!")
```

```python
# Calculate inertia for different numbers of clusters

inertias = []
k_values = range(2, 7)

for k in k_values:
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertias.append(kmeans.inertia_)

print("Inertia values:")
for k, inertia in zip(k_values, inertias):
    print(f"K = {k}: {inertia:.2f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertias,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate silhouette scores

silhouette_scores = []

for k in k_values:
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(
        X_scaled,
        labels
    )
    
    silhouette_scores.append(score)

print("Silhouette Scores:")

for k, score in zip(k_values, silhouette_scores):
    print(f"K = {k}: {score:.4f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for Different K Values")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Select K with the highest silhouette score
best_k = list(k_values)[
    np.argmax(silhouette_scores)
]

print("Selected number of clusters:", best_k)
```

```python
# Create final K-Means model

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

# Assign clusters
df["Cluster"] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")

df.head()
```

```python
# Count observations in each cluster

cluster_sizes = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

print(cluster_sizes)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    cluster_sizes.index.astype(str),
    cluster_sizes.values
)

plt.title("Number of Observations in Each Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

for cluster in sorted(df["Cluster"].unique()):
    
    cluster_data = df[
        df["Cluster"] == cluster
    ]
    
    plt.scatter(
        cluster_data["rm"],
        cluster_data["lstat"],
        label=f"Cluster {cluster}",
        alpha=0.6
    )

plt.xlabel("Average Number of Rooms (RM)")
plt.ylabel("Lower-status Population Percentage (LSTAT)")

plt.title("K-Means Clusters: RM vs LSTAT")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Calculate average feature values for each cluster

cluster_profile = df.groupby(
    "Cluster"
)[features].mean()

cluster_profile.round(2)
```

```python
print("Cluster Characteristics:")
print(cluster_profile.round(2))
```
